## Part 4 — The JOIN Trap

The operations team asks: "Can you give us total revenue per restaurant, and also show us how many reviews each restaurant has received?"

A teammate writes this query:

````python
# The teammate's query — run it and examine the result
con.execute("""
    SELECT qb_restaurants.name,
           SUM(qb_orders.total) AS total_revenue,
           COUNT(qb_reviews.review_id) AS review_count
    FROM qb_restaurants
    JOIN qb_orders ON qb_restaurants.restaurant_id = qb_orders.restaurant_id
    JOIN qb_reviews ON qb_orders.order_id = qb_reviews.order_id
    WHERE qb_orders.status = 'completed'
    GROUP BY qb_restaurants.name
    ORDER BY total_revenue DESC
""").df()
````
