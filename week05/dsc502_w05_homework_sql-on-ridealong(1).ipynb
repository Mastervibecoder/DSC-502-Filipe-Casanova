````python
# 3.2 — Drivers and total earnings (LEFT JOIN — include zero-trip drivers)
con.execute("""
    SELECT d.name,
           d.vehicle_type,
           COALESCE(SUM(t.fare), 0) AS total_earnings
    FROM drivers AS d
    LEFT JOIN trips AS t ON d.driver_id = t.driver_id
    GROUP BY d.driver_id, d.name, d.vehicle_type
    ORDER BY total_earnings DESC
""").df()
````
