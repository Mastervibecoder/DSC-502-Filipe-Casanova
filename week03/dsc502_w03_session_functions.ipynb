## Part 1 — Defining and calling a function

### Step 1 — the simplest possible function

A function is a named block of code. You define it once with `def`, then call it by name whenever you want it to run. Crucially: **defining a function doesn't run it**. Nothing happens until you call it.

````python
# Define the function — nothing happens yet
def welcome():
    print("Welcome to QuickBite!")

# Now call it — this is what runs the code
welcome()

# Call it again — same function, no rewriting needed
welcome()
welcome()
````

## Part 2 — Parameters

### Step 1 — one parameter

A no-parameter function always does exactly the same thing. Parameters let you pass values in — so the same function can produce different results depending on what you give it.

````python
# The parameter 'restaurant_name' becomes a local variable inside the function
def welcome_restaurant(restaurant_name):
    print(f"Welcome to QuickBite — now featuring {restaurant_name}!")

welcome_restaurant("The Curry House")
welcome_restaurant("Pizza Palace")
welcome_restaurant("Sushi Spot")
````

### Step 2 — multiple parameters

Functions can take more than one parameter. The order you define them in is the order you pass values when calling — so order matters.

````python
def delivery_estimate(distance_km, preparation_time):
    travel_time = distance_km * 3   # rough estimate: 3 mins per km
    total = preparation_time + travel_time
    print(f"Estimated delivery time: {total} minutes")

delivery_estimate(4, 15)    # 4km, 15 mins prep
delivery_estimate(2, 20)    # 2km, 20 mins prep

# Parameter order matters — try swapping them and see what happens
````

## Part 3 — Return values

### Step 1 — `return` vs `print`, clearly demonstrated

A function that uses `print()` produces visible output — but the value disappears once the function ends. A function that uses `return` hands the value back to whoever called it, where it can be stored, used in a calculation, or passed somewhere else. This distinction matters more than it might seem.

````python
# Version 1: print — produces output, but the value is gone
def fare_with_print(distance_km):
    fare = 2.50 + (distance_km * 1.20)
    print(fare)

result = fare_with_print(5)
print(f"Stored result: {result}")  # None — there's nothing to store
````

````python
# Version 2: return — hands the value back, now you can use it
def fare_with_return(distance_km):
    fare = 2.50 + (distance_km * 1.20)
    return fare

result = fare_with_return(5)
print(f"Stored result: {result}")           # usable
print(f"Fare for 5km: £{result:.2f}")       # formatted
print(f"With tip: £{result + 1.00:.2f}")   # used in an expression
````

### Step 2 — a function that returns a calculated value

````python
def calculate_fare(distance_km):
    return 2.50 + (distance_km * 1.20)

# Store it, use it, pass it around
fare_3km = calculate_fare(3)
fare_8km = calculate_fare(8)

print(f"3km fare: £{fare_3km:.2f}")
print(f"8km fare: £{fare_8km:.2f}")
print(f"Combined: £{fare_3km + fare_8km:.2f}")
````

## Part 4 — Default parameter values

### Worked example

Sometimes a parameter has a sensible value that works in most cases. You can set a **default** — the caller can override it if they need to, but doesn't have to. Defaults always go at the end of the parameter list.

````python
def delivery_estimate(distance_km, speed_kmh=30):
    time_mins = (distance_km / speed_kmh) * 60
    return round(time_mins)

# Using the default speed
print(delivery_estimate(5))       # 30 km/h assumed

# Overriding the default
print(delivery_estimate(5, 20))   # slower — peak hour
print(delivery_estimate(5, 45))   # faster — quiet roads
````

## Part 5 — Functions with conditionals and loops

*(Session 2)*

This is where everything from the last two weeks starts to pay off. Conditionals and loops inside functions make those functions genuinely powerful — reusable logic that makes decisions and processes collections.

### Worked example 1 — function with a conditional

````python
def free_delivery_check(order_total, threshold=25.00):
    if order_total >= threshold:
        return f"Free delivery! Your order of £{order_total:.2f} qualifies."
    else:
        shortfall = threshold - order_total
        return f"Add £{shortfall:.2f} more for free delivery."

print(free_delivery_check(31.00))
print(free_delivery_check(18.50))
print(free_delivery_check(18.50, threshold=20.00))  # different threshold
````

### Worked example 2 — function with a loop

````python
def summarise_orders(order_list):
    print("Order summary:")
    for total in order_list:
        print(f"  £{total:.2f}")

todays_orders = [12.50, 34.00, 8.75, 22.10, 47.80]
summarise_orders(todays_orders)

# Call it with a different list — same function, different data
yesterdays_orders = [9.20, 15.60, 28.00]
summarise_orders(yesterdays_orders)
````

## Part 6 — Variable scope (awareness demo)

A variable defined inside a function is **local** — it exists only for the duration of that function call. This section is just two cells to run and observe. No exercise this week — the concept will come up naturally as your programmes get more complex.

````python
# Demo 1: a local variable doesn't exist outside its function
def calculate_something():
    local_result = 42   # this only exists inside this function
    return local_result

calculate_something()

# Try to access local_result here — Python doesn't know it exists
print(local_result)   # NameError — run this, read the message
````

````python
# Demo 2: a variable defined outside a function is readable inside it,
# but modifying it inside doesn't change the original
order_count = 10

def process_orders():
    order_count = 0    # this creates a NEW local variable — doesn't touch the one above
    order_count += 5
    print(f"Inside function: {order_count}")

process_orders()
print(f"Outside function: {order_count}")   # still 10
````
