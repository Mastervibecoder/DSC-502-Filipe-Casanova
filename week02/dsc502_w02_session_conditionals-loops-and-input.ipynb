## Part 1 — Conditionals

### Step 1 — `if` on its own

The simplest conditional: check one thing, and if it's true, do something. If it's not true, nothing happens — Python just moves on.

````python
# A QuickBite restaurant is flagged if its rating drops below 3.0
restaurant_name = "The Curry House"
rating = 2.7

if rating < 3.0:
    print(f"{restaurant_name} has been flagged for review.")

# Try changing rating to 4.2 and re-running — the print disappears entirely.
# The if block runs or it doesn't. There's no other outcome yet.
````

**Comparison operators** — the tools you use to write conditions:

| Operator | Meaning | Example |
|---|---|---|
| `==` | Equal to | `rating == 5.0` |
| `!=` | Not equal to | `status != "closed"` |
| `<` | Less than | `time < 30` |
| `>` | Greater than | `total > 20` |
| `<=` | Less than or equal to | `time <= 30` |
| `>=` | Greater than or equal to | `rating >= 4.0` |

One thing to get used to: `=` and `==` are completely different. `=` assigns a value to a variable. `==` asks whether two things are equal. Using `=` where you meant `==` is one of the most common early mistakes in Python.

### Step 2 — adding `else`

A plain `if` either does something or does nothing. But sometimes you want a response in both cases — do one thing if the condition is true, something different if it's not. That's what `else` is for.

````python
# Now there's always an outcome — one branch or the other
restaurant_name = "Pizza Palace"
is_open = True

if is_open:
    print(f"{restaurant_name} is open — you can order.")
else:
    print(f"{restaurant_name} is closed right now.")

# Try changing is_open to False and re-running.
````

### Step 3 — adding `elif` for more than two cases

`if`/`else` covers two possibilities: true or not true. But what if there are three distinct cases — or four? You could nest `if`s inside `else` blocks, but it gets messy fast. `elif` ("else if") is the cleaner solution: it chains additional conditions, checked in order, stopping at the first one that's true.

````python
# Three cases — a rating can be highly rated, average, or poor
restaurant_name = "The Curry House"
rating = 4.2

if rating >= 4.5:
    print(f"{restaurant_name} is highly rated — featured in recommendations.")
elif rating >= 3.5:
    print(f"{restaurant_name} is average — shown in standard results.")
else:
    print(f"{restaurant_name} is poorly rated — flagged for review.")

# Python checks conditions top to bottom and stops at the first True.
# Try rating = 4.7, then 3.8, then 2.1 — each takes a different path.
````

## Part 2 — `for` loops

### Worked example — looping over a list

A loop lets you repeat an action for every item in a list — without copy-pasting the same line ten times.

````python
# Loop over a list of restaurant names
restaurants = ["The Curry House", "Pizza Palace", "Burger Barn", "Sushi Spot", "Taco Town"]

for restaurant in restaurants:
    print(restaurant)
````

````python
# Loop with an f-string for context
order_totals = [12.50, 34.00, 8.75, 22.10, 47.80]

for total in order_totals:
    print(f"Order total: £{total:.2f}")
````

### Worked example — while loop and the infinite loop

A `while` loop keeps running as long as its condition is true. That makes it useful when you don't know in advance how many times you need to repeat — but it also means you have to be careful.

````python
# A simple countdown
count = 5

while count > 0:
    print(f"Delivering in {count}...")
    count -= 1  # This is count = count - 1

print("Delivered!")
````
