## Exercise 3 — Sorting and groupby

1. Sort by `fare`, descending — what are the top 5 most expensive trips?
2. Sort by `distance_km`, ascending — what are the shortest trips?
3. Average fare by city — which city has the highest average fare?
4. Average rating by driver — who is the highest rated driver?
5. Count of trips by vehicle type
6. Average duration by city, sorted highest to lowest

````python
# 1
df.sort_values(by="fare", ascending=False).head(5)
````

````python
# 2
df.sort_values("distance_km", ascending=True)
````

````python
# 3
df.groupby("pickup_city")["fare"].mean().sort_values(ascending=False).round(2)
````

````python
# 4
df.groupby("driver_name")["rating"].agg(["mean", "count"]).sort_values("mean", ascending=False)
````

````python
# 5
df["vehicle_type"].value_counts()
````

````python
# 6
df.groupby("pickup_city")["duration_mins"].mean().sort_values(ascending=False).round(1)
````

## Exercise 4 — Dictionary and DataFrame creation

1. Create a dictionary summarising 4 RideAlong drivers — name, vehicle type, and a made-up average rating. Print each entry using a for loop and `.items()`.
2. Create a small DataFrame from this dictionary using `pd.DataFrame()`.
3. Add a fifth driver by creating a new entry in the dictionary, then recreate the DataFrame.

````python
# 1 — dictionary and loop
drivers = {
    "Leon Baptiste": {"vehicle_type": "Standard", "avg_rating": 4.7},
    "Mei Zhang": {"vehicle_type": "Executive", "avg_rating": 3.9},
    "Priya Sharma": {"vehicle_type": "Electric", "avg_rating": 4.3},
    "Tom Walsh": {"vehicle_type": "Executive", "avg_rating": 4.5},
}

for name, info in drivers.items():
    print(f"{name}: {info['vehicle_type']}, rating {info['avg_rating']}")
````

````python
# 2 — create DataFrame
df_drivers = pd.DataFrame.from_dict(drivers, orient="index")
df_drivers
````

````python
# 3 — add fifth driver, recreate
drivers["Aisha Kofi"] = {"vehicle_type": "Standard", "avg_rating": 3.6}
df_drivers = pd.DataFrame.from_dict(drivers, orient="index")
df_drivers
````
