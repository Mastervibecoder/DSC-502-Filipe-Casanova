## Part 3 — Filtering rows

### The boolean mask — two steps made visible

Filtering works by creating a Series of `True`/`False` values (the mask), then passing that into `df[...]` to keep only the `True` rows. Let's make both steps visible.

````python
# Step 1: create the mask — a Series of True/False
mask = df["customer_city"] == "London"
mask.head(10)
````

````python
# Step 2: apply the mask — keep only True rows
london_orders = df[mask]
london_orders.shape
````

````python
# In practice, both steps are combined into one line
london_orders = df[df["customer_city"] == "London"]
london_orders.head()
````

````python
# Numeric filter
high_value = df[df["order_total"] > 30]
print(f"Orders over £30: {high_value.shape[0]}")
````

### Multiple conditions — the parentheses rule

When combining conditions, each one must be wrapped in its own parentheses. Without them, Python's operator precedence causes a confusing error. Use `&` for AND and `|` for OR — not `and` / `or`.

````python
# Both conditions must be True — & operator, parentheses around each
london_high = df[(df["customer_city"] == "London") & (df["order_total"] > 30)]
print(f"London orders over £30: {london_high.shape[0]}")
````

## Part 5 — Aggregation with groupby

*(Session 2)*

This is the payoff. `groupby` splits your data into groups and lets you compute a summary for each one. Questions you couldn't answer by looking at 150 rows become answerable in a single line.

````python
# Average order total by cuisine type
# groupby splits the data | ["order_total"] selects the column | .mean() aggregates
df.groupby("cuisine_type")["order_total"].mean()
````

````python
# Different aggregations on the same groupby
print(df.groupby("cuisine_type")["order_total"].sum())
print()
print(df.groupby("cuisine_type")["order_total"].count())
````

````python
# Chaining: groupby then sort — highest rated cuisine first
# Note: .mean() on rating ignores NaN values automatically
df.groupby("cuisine_type")["rating"].mean().sort_values(ascending=False)
````

## Part 7 — Dictionaries and what a DataFrame actually is

Here's something worth knowing: **a DataFrame is essentially a dictionary of lists.**

The column names are dictionary keys. Each column's values are a list. That's the structure pandas builds when you load a CSV — and it's why `df["column_name"]` looks like a dictionary lookup. Because it is one.

````python
# A dictionary of lists
data = {
    "restaurant": ["Spice Garden", "Burger Barn", "Pizza Palace"],
    "rating": [4.5, 3.8, 4.2],
    "orders": [42, 31, 55]
}

# Create a DataFrame directly from a dictionary
mini_df = pd.DataFrame(data)
mini_df
````

````python
# Column names are dictionary keys
print(data.keys())

# Accessing a column from the DataFrame — same syntax as a dictionary
print(mini_df["rating"])
````
